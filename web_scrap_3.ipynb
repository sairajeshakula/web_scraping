# here we do extraction from mulitple pages using for loop

```python
mob=[]
for i in range(2,30):
    url='https://www.amazon.in/s?k=mobiles+under+30000+5g+phones&page=2&xpid=C8HXckAWrg5qE&crid=1GO94AW64J84A&qid=1742317149&sprefix=mobiles+%2Caps%2C354&ref=sr_pg_'+str(i)
    r=requests.get(url,headers=headers)
    soup=BeautifulSoup(r.text,'html.parser')
    product_name=soup.find_all('a',attrs={'class':"a-link-normal s-line-clamp-2 s-link-style a-text-normal"})
   
    for j in product_name:
        name=j.text
        mob.append(name)
df=pd.DataFrame({'mobile_brand':mob})
```

```python
df[['brand','series']]=df['mobile_brand'].str.split(' ',n=1,expand=True)
```

```python
df.drop('mobile_brand',axis=1,inplace=True)
```

```python
df['series']=df['series'].str.split('(').str[0]
```

```python
df['mobile']='smart_mobile'
```
