# HERE WE EXTRACT CONTENT FROM MULTIPLE PAGES USING FOR LOOP

```python

for i in range(2,30):
    url='https://www.amazon.in/s?k=laptops&page=2&xpid=USdpAV-5Gd_8Y&crid=1N25SN1KEIOU1&qid=1742319096&sprefix=laptop%2Caps%2C407&ref=sr_pg_'+str(i)
    r=requests.get(url,headers=headers)
    soup=BeautifulSoup(r.text,'html.parser')
    product_name=soup.find_all('a',attrs={'class':"a-link-normal s-line-clamp-2 s-link-style a-text-normal"})
   
    for j in product_name:
        name=j.text
        lap.append(name)
df=pd.DataFrame({'laptop_brand':lap})
```

```python
#df['brand']=df['laptop_brand'].str.strip().str.split('(').str[0].str.strip().str.split().str[0]
```

```python
df[['brand','series']]=df['laptop_brand'].str.split(' ',n=1,expand=True)
```

```python
df['category']='laptops'
```

```python
df.drop('laptop_brand',axis=1,inplace=True)
```

```python
df=df[df['brand']!='(Refurbished)'] 
```

```python
df=df[['category','brand','series']]
```
